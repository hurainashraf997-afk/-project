```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print(df)

print("Average Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())
print("Total Students:", df["Name"].count())
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print("Students with marks >= 50:")
print(df[df["Marks"] >= 50])

print("\nStudents with marks < 50:")
print(df[df["Marks"] < 50])
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45],
    "City": ["Karachi", "Lahore", "Karachi", "Islamabad", "Karachi"]
}

df = pd.DataFrame(data)

print(df)

print("\nStudents from Karachi:")
print(df[df["City"] == "Karachi"])
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Course": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print(df)

print("\nAverage Marks of Each Course:")
print(df.groupby("Course")["Marks"].mean())
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [80, None, 70, None, 90]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nNumber of Missing Values:")
print(df["Marks"].isnull().sum())

average_marks = df["Marks"].mean()

df["Marks"] = df["Marks"].fillna(average_marks)

print("\nCleaned DataFrame:")
print(df)
```
